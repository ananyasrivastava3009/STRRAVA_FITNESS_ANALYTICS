```python
import pandas as pd

# Load CSV files
daily_activity = pd.read_csv("../data/dailyActivity_merged.csv")
daily_calories = pd.read_csv("../data/dailyCalories_merged.csv")
daily_intensities = pd.read_csv("../data/dailyIntensities_merged.csv")
daily_steps = pd.read_csv("../data/dailySteps_merged.csv")
hourly_calories = pd.read_csv("../data/hourlyCalories_merged.csv")

# Load Excel file
heartrate = pd.read_excel("../data/heartrate_seconds_merged.xlsx")

print("All datasets loaded successfully!")
```

```python
datasets = {
    "Daily Activity": daily_activity,
    "Daily Calories": daily_calories,
    "Daily Intensities": daily_intensities,
    "Daily Steps": daily_steps,
    "Heart Rate": heartrate,
    "Hourly Calories": hourly_calories
}

for name, data in datasets.items():
    print(f"{name}: {data.shape[0]} rows, {data.shape[1]} columns")
```

```python
# STEP 14: STATISTICAL SUMMARY

for name, data in datasets.items():
    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)
    display(data.describe(include="all").T)
```

```python
# STEP 15: UNIQUE USERS

for name, data in datasets.items():
    if "Id" in data.columns:
        print(f"{name}: {data['Id'].nunique()} unique users")
    elif "ID" in data.columns:
        print(f"{name}: {data['ID'].nunique()} unique users")
    else:
        print(f"{name}: User ID column not found")
```

```python
# STEP 16: FIND DATE/TIME COLUMNS

for name, data in datasets.items():
    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    for column in data.columns:
        if "date" in column.lower() or "time" in column.lower():
            print(column)
```

```python
# ============================================
# STEP 17: DATA CLEANING
# ============================================

# Remove duplicate rows
daily_activity = daily_activity.drop_duplicates().copy()
daily_calories = daily_calories.drop_duplicates().copy()
daily_intensities = daily_intensities.drop_duplicates().copy()
daily_steps = daily_steps.drop_duplicates().copy()
heartrate = heartrate.drop_duplicates().copy()
hourly_calories = hourly_calories.drop_duplicates().copy()

print("Duplicate records removed successfully!")
```

```python
# ============================================
# STEP 18: CONVERT DATE COLUMNS
# ============================================

# Daily Activity
if "ActivityDate" in daily_activity.columns:
    daily_activity["ActivityDate"] = pd.to_datetime(
        daily_activity["ActivityDate"],
        errors="coerce"
    )

# Daily Calories
if "ActivityDay" in daily_calories.columns:
    daily_calories["ActivityDay"] = pd.to_datetime(
        daily_calories["ActivityDay"],
        errors="coerce"
    )

# Daily Intensities
if "ActivityDay" in daily_intensities.columns:
    daily_intensities["ActivityDay"] = pd.to_datetime(
        daily_intensities["ActivityDay"],
        errors="coerce"
    )

# Daily Steps
if "ActivityDay" in daily_steps.columns:
    daily_steps["ActivityDay"] = pd.to_datetime(
        daily_steps["ActivityDay"],
        errors="coerce"
    )

print("Date columns converted successfully!")
```

```python
# ============================================
# STEP 19: HANDLE MISSING VALUES
# ============================================

for name, data in datasets.items():
    print(f"\n{name}")
    print(data.isnull().sum())
```

```python
# ============================================
# STEP 21: SAVE CLEANED DATASETS
# ============================================

import os

# Create cleaned folder
os.makedirs("../data/cleaned", exist_ok=True)

# Save daily datasets as CSV
daily_activity.to_csv(
    "../data/cleaned/daily_activity_cleaned.csv",
    index=False
)

daily_calories.to_csv(
    "../data/cleaned/daily_calories_cleaned.csv",
    index=False
)

daily_intensities.to_csv(
    "../data/cleaned/daily_intensities_cleaned.csv",
    index=False
)

daily_steps.to_csv(
    "../data/cleaned/daily_steps_cleaned.csv",
    index=False
)

# Save Heart Rate as CSV because it is too large for Excel
heartrate.to_csv(
    "../data/cleaned/heartrate_cleaned.csv",
    index=False
)

# Save hourly calories
hourly_calories.to_csv(
    "../data/cleaned/hourly_calories_cleaned.csv",
    index=False
)

print("All cleaned datasets saved successfully!")
```

```python
# ============================================
# STEP 22: FINAL CLEANING CHECK
# ============================================

cleaned_datasets = {
    "Daily Activity": daily_activity,
    "Daily Calories": daily_calories,
    "Daily Intensities": daily_intensities,
    "Daily Steps": daily_steps,
    "Heart Rate": heartrate,
    "Hourly Calories": hourly_calories
}

for name, data in cleaned_datasets.items():
    print(
        f"{name}: "
        f"{data.shape[0]} rows × {data.shape[1]} columns | "
        f"Duplicates: {data.duplicated().sum()}"
    )
```

```python
daily_master = daily_activity.copy()

daily_master = daily_master.merge(
    daily_calories,
    left_on=["Id", "ActivityDate"],
    right_on=["Id", "ActivityDay"],
    how="left",
    suffixes=("", "_calories")
)

daily_master = daily_master.merge(
    daily_intensities,
    left_on=["Id", "ActivityDate"],
    right_on=["Id", "ActivityDay"],
    how="left",
    suffixes=("", "_intensity")
)

daily_master = daily_master.merge(
    daily_steps,
    left_on=["Id", "ActivityDate"],
    right_on=["Id", "ActivityDay"],
    how="left",
    suffixes=("", "_steps")
)

print("Master dataset created successfully!")
print("Shape:", daily_master.shape)

display(daily_master.head())
```

```python
heartrate.columns
```

```python
heartrate["Time"] = pd.to_datetime(heartrate["Time"], errors="coerce")

heartrate["ActivityDate"] = heartrate["Time"].dt.date

heart_daily = (
    heartrate.groupby(["Id", "ActivityDate"])["Value"]
    .agg(
        AvgHeartRate="mean",
        MaxHeartRate="max",
        MinHeartRate="min"
    )
    .reset_index()
)

heart_daily["ActivityDate"] = pd.to_datetime(heart_daily["ActivityDate"])

print("Daily heart-rate data created successfully!")
print("Shape:", heart_daily.shape)

display(heart_daily.head())
```

```python
daily_master["ActivityDate"] = pd.to_datetime(
    daily_master["ActivityDate"],
    errors="coerce"
)

daily_master = daily_master.merge(
    heart_daily,
    on=["Id", "ActivityDate"],
    how="left"
)

print("Heart-rate data merged successfully!")
print("New shape:", daily_master.shape)

display(daily_master.head())
```

```python
query = """
SELECT *
FROM fitness_data
LIMIT 10;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT COUNT(DISTINCT Id) AS Total_Users
FROM fitness_data;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT
    ROUND(AVG(TotalSteps), 2) AS Avg_Steps,
    ROUND(AVG(TotalDistance), 2) AS Avg_Distance,
    ROUND(AVG(Calories), 2) AS Avg_Calories,
    ROUND(AVG(VeryActiveMinutes), 2) AS Avg_VeryActive_Minutes,
    ROUND(AVG(SedentaryMinutes), 2) AS Avg_Sedentary_Minutes
FROM fitness_data;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT
    Id,
    ActivityDate,
    TotalSteps,
    TotalDistance,
    Calories
FROM fitness_data
ORDER BY TotalSteps DESC
LIMIT 10;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT
    strftime('%w', ActivityDate) AS Day_Number,
    ROUND(AVG(TotalSteps), 2) AS Avg_Steps,
    ROUND(AVG(Calories), 2) AS Avg_Calories
FROM fitness_data
GROUP BY Day_Number
ORDER BY Day_Number;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT
    ROUND(AVG(AvgHeartRate), 2) AS Average_Heart_Rate,
    ROUND(MAX(MaxHeartRate), 2) AS Maximum_Heart_Rate,
    ROUND(MIN(MinHeartRate), 2) AS Minimum_Heart_Rate
FROM fitness_data
WHERE AvgHeartRate IS NOT NULL;
"""

display(pd.read_sql_query(query, conn))
```
